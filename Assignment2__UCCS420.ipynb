```python
#Q3
#(i)Generate a list of 100 random numbers between 100 and 900 (inclusive).
import random
random.seed(1024160086)
nums = [random.randint(100, 900) for _ in range(100)]
print(f"Random nums:{nums}")
```

```python
#(ii)Count and print all odd numbers in the list.
odd_count = sum(1 for num in nums if num % 2 != 0)
print("\nNumber of odd numbers:", odd_count)
```

```python
#(iii)Count and print all even numbers in the list.
even_count = sum(1 for num in nums if num % 2 == 0)
print("Number of even numbers:", even_count)
```

```python
#(iv)Count and print all prime numbers in the list, and additionally build the actual list of prime numbers found using list comprehension.
def is_prime(n):
    if n < 2:
        return False

    for i in range(2, int(n ** 0.5) + 1):
        if n % i == 0:
            return False

    return True
prime_no = [num for num in nums if is_prime(num)]
print("prime no",prime_no)
print("Number of prime numbers:", len(prime_no))
```

```python
#(v)Print the number that occurs most frequently in your list, and how many times it occurs.
freq = {}
for num in nums:
    freq[num] = freq.get(num, 0) + 1
most_frequent = max(freq, key=freq.get)
print("\nMost frequent number:", most_frequent,"frequency =",freq[most_frequent])
```

```python
#(vii) Find the intersection of A and B
intersection_set=A.intersection(B)
print("intersection of A and B",intersection_set)
```

```python
#(viii) Find A − B and B − A
A_minus_B = A.difference(B)
B_minus_A = B.difference(A)
print("A - B:", A_minus_B)
print("B - A:", B_minus_A)
#Difference removes elements of one set from another, whereas symmetric_difference returns elements that are in either set but not in both.
```

```python
#(ix)Find the symmetric difference of A and B.
sym_diff = A.symmetric_difference(B)
print("symmetric difference:",sym_diff)
```

Q.6 Create a sample dataset (employees.csv) containing information about employees in a company. WAP to determine the following using dataset

```python
# a) Shape (number of rows and columns) of the DataFrame
df = pd.read_csv("employee.csv")
print(df.shape)
```

```python
# b) Summary of the DataFrame that includes the data types and non-null counts for each column.
df.info()
```

```python
# c) Generate descriptive statistics.
print(df.describe())
```

```python
# d) Display the first 5 rows and last 3 rows of the dataset.
print(df.head(5))
print(df.tail(3))
```

```python
# e) Calculate the following statistics from the dataset:
# i. Average salary
print("Average Salary:",df["Salary"].mean())
# ii. Total bonus
print("Total bonus",df["Bonus"].sum())
# iii. Youngest employee's age
print("Youngest Employee Age:",df["Age"].min())
# iv. Highest performance rating
print("Highest Performance Rating:",df["Rating"].max())
```

```python
# f) Sort by Salary in descending order
sorted_df = df.sort_values(by="Salary", ascending=False)
print("Salary in Descending Order:",sorted_df)
```

```python
# g) Add Category column based on Rating
def categorize_rating(rating):
    if rating >= 4.5:
        return "Excellent"
    elif rating >= 4.0:
        return "Good"
    else:
        return "Average"
df["Category"] = df["Rating"].apply(categorize_rating)
print(df[["Name", "Rating", "Category"]])
```

```python
# h) Identify missing values
print(df.isnull().sum())
```

```python
# i) Rename Employee_ID to ID
df.rename(columns={"Employee_ID": "ID"}, inplace=True)
print(df)
```

```python
# j) Find employees

# i. More than 5 years of experience
print("empoyees with experience >5 years")
print(df[df["Years_of_Experience"] > 5])

# ii. Belong to IT department
print("empoyees that Belong to IT department")
print(df[df["Department"] == "IT"])
```

```python
# k) Add Tax column (10% of Salary)
df["Tax"] = df["Salary"] * 0.10
print("Tax column added:")
print(df)
```

```python
# l) Save modified DataFrame to new CSV
df.to_csv("employees_modified.csv", index=False)
print(" Modified dataset saved as employees_modified.csv")
```

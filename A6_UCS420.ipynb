```python
temperature = np.array([25,28,31,35,27,33,40])
#(a)
temperature = np.array([25, 28, 31, 35, 38, 27, 33, 40])
corrected = temperature + 2
print("Original temperatures:", temperature)
print("Corrected temperatures:", corrected)
#(b)
fahrenhite = ((9/5)*corrected)+ 32
print("Temperature in fahrenhite",fahrenhite)
#(c)
print("Readings greater than 32:",corrected[corrected>32])
#(d)
count = np.sum(corrected>32)
print("number of reading greater than 32:",count)
#(e)
print("""
Vectorization performs operations on the whole array at once, making the code faster and shorter.
Boolean indexing quickly filters values based on a condition without using a for loop.
""")
```

Q2.

```python
steps = np.array([[5000,6200,7100],[8000,7500,9000],[4500,5100,4800],[9000,8500,9500]])
#(a)
total_steps = np.sum(steps)
print("Total:",total_steps)
#(b)
mean_steps = np.mean(steps)
print("Mean",mean_steps)
#(c)
max_steps = np.max(steps)
min_steps = np.min(steps)
print("max",max_steps)
print("min",min_steps)
#(d)
total_steps_axis0 = np.sum(steps,axis=0)
print("Daily totals:",total_steps_axis0)
#(e)
total_steps_axis1 = np.sum(steps,axis=1)
print("User totals:",total_steps_axis1)
#(f)
print("Maximum position:", np.unravel_index(np.argmax(steps), steps.shape))
```

```python
#(a)
original = np.array([1,2,3,4,5,6])
print("original:",original)
#(b)
subset = original[1:5]
print("subset:",subset)
#(c)
subset[0] = 999
print("original",original)
print("subset",subset)
#(d)
copied = original[1:5].copy()
copied[0] = 500
print("Original:", original)
print("Copied:", copied)
#(e)
arr = np.arange(1, 13).reshape(3, 4)
print("arr:\n", arr)
#(f)
print("First row:", arr[0])
print("Last row:", arr[-1])
print("Second column:", arr[:, 1])
print("Rows 1-2, Columns 2-3:\n", arr[0:2, 1:3])
#(g)
flat = arr.flatten()
rav = arr.ravel()
print("Flatten:", flat)
print("Ravel:", rav)
#(h)
rav[0] = 100
print("After modifying ravel:", arr)
#(i)
flat[1] = 200
print("After modifying flatten:", arr)
#(j)
print("Shape:", arr.shape)
print("Dimensions:", arr.ndim)
print("Size:", arr.size)
print("Data type:", arr.dtype)
```

Q4.

```python
y = np.array([40, 65, 30, 85])
X = np.array([[6, 70, 3],[5, 50, 6],[8, 80, 2],[4, 30, 8]])
#(a)
print("Shape:", X.shape)
print("Dimensions:", X.ndim)
#(b)
print("X.T:\n", X.T)
#(c)
prod = X.T @ X
print("X.T @ X:\n", prod)
#(d)
inverse = np.linalg.inv(prod)
print("Inverse:\n", inverse)
#(e)
beta = inverse @ X.T @ y
print("Coefficients:", beta)
#(f)
print("Sleep coefficient:", beta[0])
print("Activity coefficient:", beta[1])
print("Stress coefficient:", beta[2])
print("""
The coefficients show how each feature affects the assistance score.
A positive coefficient increases the score, while a negative coefficient decreases it.
""")
#(g)
new_user = np.array([5, 40, 7])
prediction = new_user @ beta
print("Predicted assistance score:", prediction)
```

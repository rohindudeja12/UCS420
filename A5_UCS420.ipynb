```python
#(b)Find the most frequent value and their indice(s) in the following arrays:
#(i)
x = np.array([1,2,3,4,5,1,2,1,1,1])
frequency_x = np.bincount(x).argmax()
index= np.where(x == frequency_x)[0]
print(f"The most frequent value is {frequency_x} and its index is {index}")
```

```python
#(ii)
y = np.array([1,1,1,2,3,4,2,4,3,3])
frequency_y = np.bincount(y).argmax()
index_y = np.where(y==frequency_y)[0]
print(f"The most frequent value is {frequency_y} and its index is {index_y}")
```

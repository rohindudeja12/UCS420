 Q2. Generate and Score a Hypothesis - Implement a scoring function that takes a query string and returns all matching entries ranked by confidence

```python
def score_query(query, df):
    query_words = set(query.lower().split())
    results = []

    for _, row in df.iterrows():
        keywords = set(row["keywords"].lower().split())
        score = len(query_words.intersection(keywords))

        if score > 0:
            results.append({
                "question": row["question"],
                "answer": row["answer"],
                "category": row["category"],
                "score": score
            })

    results.sort(key=lambda x: x["score"], reverse=True)

    return results
query = input("Enter your query: ")
results = score_query(query, df)
for result in results:
    print(result)
```

Q6.  Modify your Q2 scoring function so that if two or more entries tie for the highest score, it does not silently pick one — it prints all matching entries instead, so the user can see every equally good match. Demonstrate with one query that produces a tie (e.g. a query matching both "fee" entries) and one that doesn't.

```python
def score_hypothesis(query, df):

    query_words = set(query.lower().split())

    results = []

    for _, row in df.iterrows():

        words = set(
            row["question"].lower().split() +
            row["keywords"].lower().split()
        )

        matches = query_words & words

        confidence = len(matches) / len(query_words)

        if confidence > 0:
            results.append({
                "question": row["question"],
                "answer": row["answer"],
                "category": row["category"],
                "confidence": round(confidence, 2)
            })

    if not results:
        print("No matching entries found.")
        return

    max_score = max(r["confidence"] for r in results)

    best_matches = [
        r for r in results
        if r["confidence"] == max_score
    ]

    print("\nHighest confidence:", max_score)

    if len(best_matches) > 1:
        print("Tie found. ")
    else:
        print("Best match:")

    print(pd.DataFrame(best_matches))
print("\n Testing Tie: ")
score_hypothesis("fee", df)

print("\n Testing No Tie: ")
score_hypothesis("password", df)

```

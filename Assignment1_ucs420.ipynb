Math Library

```python
import math as m
print ("exp(3) -> ", m.exp(3))
print ("log(100,2) -> ", m.log(100,2))
print ("log(100,10) -> ", m.log(100,10))
print ("log10(100) -> ", m.log10(100))
print ("m.cos(30)  -> ", m.cos(30))
print ("m.sin(30)  -> ", m.sin(30))
print ("m.tan(30)  -> ", m.tan(30))
print ("m.sqrt(81) -> ", m.sqrt(81))
print ("m.ceil(52.3)-> ", m.ceil(52.3))
print ("m.floor(79.9)--> ", m.floor(79.9))
```

Strings

```python
var = 'Hello World!'
print ("var      --> ", var)
print ("var[0]   --> ", var[0])
print ("var[1:5] --> ", var[1:5])
print ("var[:-5] --> ", var[:-5])
```

```python
var = 'HEY from Rohin'
print ("String --> ", var)
print ("Length --> : ", len(var))
print ("Upper  --> : ", var.upper())
print ("Lower  --> : ", var.lower())
```

String split

```python
var =" Indian,   Army    "

print("String    --> ", var)
print("Length    --> ", len(var))
print("var split --> ", var.split())
print("var split --> ", var.split(' '))
print("var split --> ", var.split(','))

# Strip + Split
print("var split --> ", var.strip().split(','))
```

Count in string

```python
var=" Indian Army    "
print ("String       --> ", var)
print ("Count of ' ' --> ", var.count(' '))
print ("Count of 'a' --> ", var.count('a'))
print ("Count of 'an' --> ", var.count('an'))
```
